### Univariate Analysis - Single Variable (Column) Analysis

- Univariate analysis means analysis on a single column

#### Steps for doing Univariate Analysis on Numerical Columns
- **Performing Descriptive Stats** -> Compute basic summary statistics such as mean,median,mode,standard deviation,range and quartile because these statistics give us general understanding of distribution of data and also identify skewness and outliers

- **Visualizations** - Using Visualizations to explore the distribution of data. *Some common visualizations for numerical columns include histograms, box plots and density plots*

- **Identify Outliers** - Using Visualizations we identify outliers.

- **Skewness** - It is a statistical measure that describes the asymmetry of a data distribution from a normal, bell-curve shape. Check for skewness in data and transform the data using statistical methods 

** **

#### Univariate Analysis On Numerical Columns

#### Analysis On Age Column

```python
df['Age']
```

```python
df['Age'].describe()
#50% is taken as median i.e 50% of people are less than or equal to age == 28 (same goes for 25 and 75)
```

```python
df['Age'].plot(kind='hist',bins=20)
```

```python
df['Age'].plot(kind='kde')
```

```python
df['Age'].skew()
```

- *if result is near zero then it is perfect normal distribution*

- *if result is nearby zero then it is almost normal distribution*

- *if result is more positive that has high forward distance from zero then it is positive skewness (in simple terms)*

- *if result is more negative that has high backward distance from zero then it is negative skewness (in simple terms)*

```python
df['Age'].plot(kind='box')
```

```python
df[df['Age']>65]
```

```python
df['Age'].isnull().sum()
```

```python
df['Age'].isnull().sum()/len(df['Age'])
# nearly 20% values are missing (rounded off to 20)
```

```python
df[df['Age'].isnull()==True]
```

```python
df.head(10)
```

**Conclusions**

- *The Age Column is almost normally distributed*

- *Nearly 20% percentage of the values are missing*

### Univariate Analysis on Categorical columns
- **Descriptive Statistics** ->  Compute the frequency Distribution of the categories of the column

- **Visualizations** -> Using Visualizations such as count plots and pie charts to explore the distribution of categories

#### Analysis on Survived Column

- *This is Categorical Column because only has two types of data*

- *Either 1 or 0*

```python
df['Survived']
```

```python
df['Survived'].value_counts()
#0 means death 1 means survived
```

```python
df['Survived'].value_counts().plot(kind='bar',color=['darkblue','darkred'])
```

```python
#verifying in terms of percentage how many people died
df['Survived'].value_counts().plot(kind='pie',autopct='%0.1f%%',colors=['red','skyblue'])
```

*Note - More than 60% of people are dead who boarded the titanic*

```python
df['Survived'].isnull().sum()
```

**Conclusions**

- *Less than 40% of people who boarded the titanic are alive so the incident was severe*

- *Out of 891 people 549 (61.6%) are dead and remaining 342 (38.4%) are alive*

- *Note - There are no null values i.e missing values*
