```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Steps on Bivariate Analysis

*Analysis on Two Columns*

- **Step-1** -> *Selecting two columns*

- **Step-2** -> *Understanding the relationship between those two columns*

  - *Both the Columns could be* **Numerical & Numerical**
 
    - *We can plot scatterplots(regression plots),2D hist plots & 2D KDEplots to visualize and understand the relationship*
   
    - *We can also check correlation coefficient to understand if the relation is linear relationship or other relationship*
 
  - *Both the Columns could be* **Numerical & Categorical**
 
    - *Create Visualization that compare the distribution of numerical data across the categories of the categorical data*
   
    - *We can plot graphs like barplot, boxplot, kdeplots, violinplots and even sometimes scatterplots*
 
  - *Both the Columns could be* **Categorical & Categorical**
 
    - *We can create Cross tabulations or Contingency tables that show the distribution of values in one categorical column grouped by the values in other categorical column*
   
    - *We can visualize by plotiing heatmaps, treemaps, stacked barplots*
   
- **Step-3** -> *Do some feature engineering and draw conclusions*

### *Bi-variate analysis the Survived and Pclass Columns*

```python
df[['Survived','Pclass']]
```

```python
pd.crosstab(df['Survived'],df['Pclass'],normalize='columns')*100
#on the basis of index finding out what percentage of people survived and not survived
```

*Note - 62.9% out of 100% passengers survived in Pclass '1', 47.2% out of 100% passengers survived in Pclass '2', 24.2% out of 100% passengers survived in Pclass '1'*

```python
sns.heatmap(pd.crosstab(df['Survived'],df['Pclass'],normalize='columns')*100)
```

**Conclusion**

- *1 means survived and 0 means not survived*

- *More than 60% of the Pclass '1' passengers survived and less than 25% of Pclass '3' passengers survived*

- *So passengers travelling in Pclass '1' had a more chance of surviving than passengers travelling in Pclass '3'*

##### *2. First find the Total Family Size*

```python
DF['Family_Size'] = (DF['SibSp']+DF['Parch']+1)
```

```python
DF
```

```python
DF[DF['Ticket']=='CA. 2343']
```

*Creating a new column 'Family_Type' using the 'Family_Size column'*
- *If 1 then Alone*
- *If 2-4 then Small*
- *If >=5 then Large*

```python
def transform_family_size(num):
    if num==1:
        return 'alone'
    elif num>1 and num<5:
        return 'small'
    else:
        return 'large'
```

```python
DF['Family_Type'] = DF['Family_Size'].apply(transform_family_size)
```

```python
DF
```

#### *Bivariate Analysis on Survived and Family_Type columns*

```python
sns.heatmap(pd.crosstab(DF['Survived'],DF['Family_Type'],normalize='columns')*100)
```

- *So chances of not surviving in Titanic if a passenger is alone is 69.6% out of 100%*

- *So chances of not surviving in Titanic if a passenger is with small family is 42.1% out of 100%*

- *So chances of not surviving in Titanic if a passenger is with large family is 83.8% out of 100%*
